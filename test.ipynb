```python
# PART 1 : IMPORT LIBRARIES
import random
import sqlite3
from datetime import datetime, timedelta

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# PART 2 : CLASS MEMBER
class Member:
    def __init__(
        self,
        member_id,
        name,
        age,
        weight_kg,
        is_vip=False
        ):
        self.member_id = member_id
        self.name = name
        self.age = age
        self.weight_kg = weight_kg
        self.is_vip = is_vip
    def get_member_type(self):
        if self.is_vip:
            return "VIP"
        return "General"
    def __str__(self):
        return (
            f"Member ID: {self.member_id} | "
            f"Name: {self.name} | "
            f"Age: {self.age} | "
            f"Weight: {self.weight_kg} kg | "
            f"Type: {self.get_member_type()}"
        )
```
